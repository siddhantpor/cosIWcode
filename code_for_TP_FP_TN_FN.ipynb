```python
import numpy as np
from sklearn.metrics import f1_score, precision_score, recall_score, roc_auc_score

# this notebook contains code for evaluation data visualization

# true positive
TP = 44
FP = 2
FN = 248
TN = 4706

# evaluation metrics - important
sensitivity = TP / (TP + FN)
specificity = TN / (TN + FP)
precision   = TP / (TP + FP)
recall      = sensitivity
accuracy    = (TP + TN) / (TP + TN + FP + FN)
f1          = 2 * (precision * recall) / (precision + recall)

print("Sensitivity (Recall):", sensitivity)
print("Specificity:", specificity)
print("Precision:", precision)
print("Accuracy:", accuracy)
print("F1 Score:", f1)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm


font_path = "lmsans10-regular.otf"  # font for theme
font = fm.FontProperties(fname=font_path)
plt.rcParams['font.family'] = font.get_name()

# rates
TP = 44 / (44 + 2)
FP = 2 / (44 + 2)
FN = 248 / (4706 + 248)
TN = 4706 / (4706 + 248)

cm = np.array([
    [TP, FN],
    [FP, TN]
])


fig, ax = plt.subplots(figsize=(6, 6))
im = ax.imshow(cm, cmap="Paired")


ax.set_xlabel("Predicted", fontproperties=font)
ax.set_ylabel("Actual", fontproperties=font)
ax.set_title("Confusion Matrix", fontproperties=font)
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(["Positive", "Negative"], fontproperties=font)
ax.set_yticklabels(["Positive", "Negative"], fontproperties=font)

# use percentage for better visual effect
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        percent_text = f"{cm[i, j]*100:.1f}%"
        color = "white" if cm[i, j] > 0.5 else "black"
        ax.text(j, i, percent_text, ha="center", va="center", color=color, fontproperties=font)

plt.tight_layout()
plt.show()

# code from last cell
sensitivity = TP / (TP + FN)
specificity = TN / (TN + FP)
precision   = TP / (TP + FP)
recall      = sensitivity
accuracy    = (TP + TN) / (TP + TN + FP + FN)
f1          = 2 * (precision * recall) / (precision + recall)

print("Sensitivity (Recall):", round(sensitivity, 4))
print("Specificity:", round(specificity, 4))
print("Precision:", round(precision, 4))
print("Accuracy:", round(accuracy, 4))
print("F1 Score:", round(f1, 4))
```
